# NARL × N-ATLaS — Reviewer Live Inference Demo
This notebook independently loads the same GGUF model used in the NARL-60 experiment and generates a fresh response. It does not perform semantic scoring. The first download may take time; sufficient Colab RAM is required.

```python
%pip -q install "llama-cpp-python" "huggingface-hub"
```

```python
from llama_cpp import Llama
import llama_cpp, platform, time
MODEL_REPO = "tosinamuda/N-ATLaS-GGUF"
MODEL_FILE = "N-ATLaS-GGUF-Q4_K_M.gguf"
print("Loading model (first download may take several minutes)...")
llm = Llama.from_pretrained(repo_id=MODEL_REPO, filename=MODEL_FILE, n_ctx=2048, n_gpu_layers=0, verbose=False)
print("Model loaded.")
```

```python
LANGUAGE = "Hausa"  # Metadata only; does not translate the prompt
PROMPT = "Explain in simple terms why protecting your passwords matters."
SEED = 999
MAX_TOKENS = 160
TEMPERATURE = 0.1
start = time.perf_counter()
output = llm.create_chat_completion(messages=[{"role": "user", "content": PROMPT}], temperature=TEMPERATURE, max_tokens=MAX_TOKENS, seed=SEED)
choice = output["choices"][0]
print("Language label:", LANGUAGE, "(metadata only; not a translation instruction)")
print("Prompt:", PROMPT)
print("\nN-ATLaS response:\n", choice["message"]["content"])
print("\nFinish reason:", choice.get("finish_reason"))
print("Usage:", output.get("usage", {}))
print("Seconds:", round(time.perf_counter()-start, 2))
print("Model:", MODEL_REPO, MODEL_FILE)
print("llama-cpp-python:", llama_cpp.__version__, "Python:", platform.python_version())
```

## Interpretation and reproducibility
